```python
def flatten_dict(data, parent_key='', sep='_'):
    """
    Recursively flatten nested dictionaries.

    Example:
        {'BusinessParking': {'garage': True, 'street': False}}

    becomes:
        {
            'BusinessParking_garage': True,
            'BusinessParking_street': False
        }
    """

    flattened = {}

    if not isinstance(data, dict):
        return flattened

    for key, value in data.items():

        # Create the new column name
        new_key = f"{parent_key}{sep}{key}" if parent_key else str(key)

        # If the value is another dictionary, recursively flatten it
        if isinstance(value, dict):
            flattened.update(
                flatten_dict(value, new_key, sep)
            )

        # Otherwise store the value
        else:
            flattened[new_key] = value

    return flattened
```

```python
attributes_flattened = business_df['attributes'].apply(
    lambda x: flatten_dict(x)
)
```

```python
attributes_df = pd.DataFrame(
    attributes_flattened.tolist(),
    index=business_df.index
)
```

```python
print(attributes_df.shape)
print(attributes_df.columns.tolist())
```

```python
hours_flattened = business_df['hours'].apply(
    lambda x: flatten_dict(x)
)

hours_df = pd.DataFrame(
    hours_flattened.tolist(),
    index=business_df.index
)
hours_df = hours_df.rename(
    columns={
        column: f"hours_{column}"
        for column in hours_df.columns
    }
)
```

```python
business_df = pd.concat(
    [
        business_df,
        attributes_df, 
        hours_df
    ],
    axis=1
)
```

```python
business_df.head(10)
```

```python
business_df['categories'] = business_df['categories'].apply(
    lambda x: [category.strip() for category in x.split(',')]
    if pd.notna(x) else []
)
```

```python
business_df = business_df.fillna('Unknown')
```

```python

def process_elite(value):
    if pd.isna(value) or value == '' or value == 'None':
        return pd.Series({
            'elite_years_count': 0,
            'elite_first_year': 'Unknown',
            'elite_last_year': 'Unknown',
            'is_elite': False
        })

    years = [
        int(year.strip())
        for year in str(value).split(',')
        if year.strip().isdigit()
    ]

    if not years:
        return pd.Series({
            'elite_years_count': 0,
            'elite_first_year': 'Unknown',
            'elite_last_year': 'Unknown',
            'is_elite': False
        })

    return pd.Series({
        'elite_years_count': len(years),
        'elite_first_year': min(years),
        'elite_last_year': max(years),
        'is_elite': True
    })


elite_df = user_df['elite'].apply(process_elite)

user_df = pd.concat(
    [
        user_df.drop(columns=['elite']),
        elite_df
    ],
    axis=1
)
```
